# Data Train-Test 

```python
x= data[['Avg. Session Length', 'Time on App',
       'Time on Website', 'Length of Membership']]
y= data['Yearly Amount Spent']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=.25, random_state=232)
```

```python

lm = LinearRegression()

lm.fit(x_train, y_train)
```

```python
prediction = lm.predict(x_test)
```

```python
metrics.explained_variance_score(y_test, prediction)
```

```python
coffecient = pd.DataFrame(lm.coef_, x.columns, columns=['Coff'])
coffecient
```

```python
print("Intercept (b0): {:.4f}".format(lm.intercept_))
print()

equation = "Yearly Amount Spent = {:.2f}".format(lm.intercept_)
for name, coef in zip(x.columns, lm.coef_):
    sign = "+" if coef >= 0 else "-"
    equation += "\n    {} {:.2f} * {}".format(sign, abs(coef), name)
print(equation)

coefficient_table = pd.DataFrame({'Coefficient': lm.coef_}, index=x.columns)
coefficient_table.loc['Intercept'] = lm.intercept_
coefficient_table
```

```python
def regression_metrics(y_true, y_pred, n_features):
    n = len(y_true)
    mse = metrics.mean_squared_error(y_true, y_pred)
    r2 = metrics.r2_score(y_true, y_pred)
    return {
        'MAE': metrics.mean_absolute_error(y_true, y_pred),
        'MSE': mse,
        'RMSE': np.sqrt(mse),
        'MAPE (%)': np.mean(np.abs((y_true - y_pred) / y_true)) * 100,
        'R2': r2,
        'Adjusted R2': 1 - (1 - r2) * (n - 1) / (n - n_features - 1),
        'Explained Variance': metrics.explained_variance_score(y_true, y_pred),
    }

train_prediction = lm.predict(x_train)

metrics_table = pd.DataFrame({
    'Train': regression_metrics(y_train, train_prediction, x.shape[1]),
    'Test': regression_metrics(y_test, prediction, x.shape[1]),
})
metrics_table.round(4)
```

# Model Visualising

```python
from scipy import stats

residuals = y_test - prediction
test_r2 = metrics.r2_score(y_test, prediction)
test_rmse = np.sqrt(metrics.mean_squared_error(y_test, prediction))

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Actual vs Predicted
ax = axes[0, 0]
ax.scatter(y_test, prediction, alpha=0.7, edgecolor='white')
limits = [min(y_test.min(), prediction.min()), max(y_test.max(), prediction.max())]
ax.plot(limits, limits, 'r--', label='Perfect prediction')
ax.set_xlabel('Actual Yearly Amount Spent ($)')
ax.set_ylabel('Predicted Yearly Amount Spent ($)')
ax.set_title('Actual vs Predicted  (R² = {:.4f}, RMSE = {:.2f})'.format(test_r2, test_rmse))
ax.legend()

# 2. Residuals vs Predicted
ax = axes[0, 1]
ax.scatter(prediction, residuals, alpha=0.7, edgecolor='white')
ax.axhline(0, color='r', linestyle='--')
ax.set_xlabel('Predicted Yearly Amount Spent ($)')
ax.set_ylabel('Residual ($)')
ax.set_title('Residuals vs Predicted')

# 3. Residual distribution
ax = axes[1, 0]
sb.histplot(residuals, bins=30, kde=True, ax=ax)
ax.axvline(0, color='r', linestyle='--')
ax.set_xlabel('Residual ($)')
ax.set_title('Residual Distribution  (mean = {:.2f})'.format(residuals.mean()))

# 4. Q-Q plot of residuals
ax = axes[1, 1]
stats.probplot(residuals, dist='norm', plot=ax)
ax.set_title('Q-Q Plot of Residuals')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 5. Coefficients
ax = axes[0]
sorted_coef = pd.Series(lm.coef_, index=x.columns).sort_values()
bars = ax.barh(sorted_coef.index, sorted_coef.values, color='steelblue')
for bar, value in zip(bars, sorted_coef.values):
    ax.text(value + 0.5, bar.get_y() + bar.get_height() / 2, '${:.2f}'.format(value), va='center')
ax.set_xlabel('Change in Yearly Amount Spent ($) per 1 unit increase')
ax.set_title('Model Coefficients  (Intercept = {:.2f})'.format(lm.intercept_))
ax.set_xlim(0, sorted_coef.max() * 1.15)

# 6. Correlation heatmap
ax = axes[1]
sb.heatmap(data.select_dtypes(include='number').corr(), annot=True, fmt='.2f',
           cmap='coolwarm', vmin=-1, vmax=1, square=True, ax=ax)
ax.set_title('Correlation Heatmap')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

error_metrics = metrics_table.loc[['MAE', 'RMSE', 'MAPE (%)']]
error_metrics.plot(kind='bar', ax=axes[0], rot=0)
axes[0].set_title('Error Metrics (lower is better)')
axes[0].set_ylabel('Value')

score_metrics = metrics_table.loc[['R2', 'Adjusted R2', 'Explained Variance']]
score_metrics.plot(kind='bar', ax=axes[1], rot=0)
axes[1].set_title('Score Metrics (higher is better)')
axes[1].set_ylim(0.95, 1.0)

for ax in axes:
    for container in ax.containers:
        ax.bar_label(container, fmt='%.3f', fontsize=8)

plt.tight_layout()
plt.show()
```
