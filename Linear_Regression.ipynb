# Data Train-Test 

```python
x= data[['Avg. Session Length', 'Time on App',
       'Time on Website', 'Length of Membership']]
y= data['Yearly Amount Spent']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=.25, random_state=232)
```

```python

lm = LinearRegression()

lm.fit(x_train, y_train)
```

```python
prediction = lm.predict(x_test)
```

```python

sb.distplot((y_test-prediction),bins=50);
```

```python
metrics.explained_variance_score(y_test, prediction)
```

```python
coffecient = pd.DataFrame(lm.coef_, x.columns, columns=['Coff'])
coffecient
```

```python
print("Intercept (b0): {:.4f}".format(lm.intercept_))
print()

equation = "Yearly Amount Spent = {:.2f}".format(lm.intercept_)
for name, coef in zip(x.columns, lm.coef_):
    sign = "+" if coef >= 0 else "-"
    equation += "\n    {} {:.2f} * {}".format(sign, abs(coef), name)
print(equation)

coefficient_table = pd.DataFrame({'Coefficient': lm.coef_}, index=x.columns)
coefficient_table.loc['Intercept'] = lm.intercept_
coefficient_table
```

```python
def regression_metrics(y_true, y_pred, n_features):
    n = len(y_true)
    mse = metrics.mean_squared_error(y_true, y_pred)
    r2 = metrics.r2_score(y_true, y_pred)
    return {
        'MAE': metrics.mean_absolute_error(y_true, y_pred),
        'MSE': mse,
        'RMSE': np.sqrt(mse),
        'MAPE (%)': np.mean(np.abs((y_true - y_pred) / y_true)) * 100,
        'R2': r2,
        'Adjusted R2': 1 - (1 - r2) * (n - 1) / (n - n_features - 1),
        'Explained Variance': metrics.explained_variance_score(y_true, y_pred),
    }

train_prediction = lm.predict(x_train)

metrics_table = pd.DataFrame({
    'Train': regression_metrics(y_train, train_prediction, x.shape[1]),
    'Test': regression_metrics(y_test, prediction, x.shape[1]),
})
metrics_table.round(4)
```
